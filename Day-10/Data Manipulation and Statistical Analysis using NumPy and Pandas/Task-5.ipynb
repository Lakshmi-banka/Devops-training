```python
df["Energy_Per_Person"] = df["Electricity_kWh"] / df["Occupancy"]
```

```python
df
```

```python
df[["Building_ID", "Occupancy","Electricity_kWh", "Energy_Per_Person"]]
```

```python
df["Net_Electricity_kWh"] = df["Electricity_kWh"] - df["Solar_Energy_kWh"]
```

```python
df
```

```python
net = df["Net_Electricity_kWh"]

print("Mean:", net.mean())
print("Median:", net.median())
print("Standard deviation:", net.std())
print("Minimum:", net.min())
print("Maximum:", net.max())
print("Total:", net.sum())
print("Variance:", net.var())
print("Count:", net.count())
```

```python
print(df["Net_Electricity_kWh"].std(ddof=0))
```

```python
df["Energy_Category"] = pd.cut(
    df["Electricity_kWh"],
    bins=[-np.inf, 80, 110, np.inf],
    labels=["Low", "Medium", "High"]
)

df[["Building_ID", "Electricity_kWh",
          "Energy_Category"]]
```

```python
high_energy = df[df["Electricity_kWh"] > 100]

high_energy
```

```python
high_occupancy = df[df["Occupancy"] > 100]

high_occupancy
```

```python
labs = df[df["Building_Type"] == "Computer Lab"]

labs
```

```python
result = df[
    (df["Electricity_kWh"] > 100) &
    (df["Solar_Energy_kWh"] > 30)
]

result
```

```python
summary = df.groupby("Building_Type", observed=True).agg(
    Average_Electricity=("Electricity_kWh", "mean"),
    Total_Electricity=("Electricity_kWh", "sum"),
    Average_Occupancy=("Occupancy", "mean"),
    Building_Count=("Building_ID", "count")
)

summary
```

```python
df_sorted = df.sort_values(
    by="Electricity_kWh",
    ascending=False
)

df_sorted
```
